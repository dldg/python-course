```python
import os
import re
```

```python
root_folder = 'instructions'
```

```python
pattern = r'\d{3}-\d{3}-\d{4}'
```

```python
for folder, sub_folders,files in os.walk(root_folder):
    for file in files:
        file_path = folder + "/" + file
        f = open(file_path,'r+')
        # read all lines in a list
        lines = f.readlines()
        for line in lines:
            # check if string present on a current line
            phone = re.search(pattern,line)
            if phone:
                print(phone.group())
        f.close()
```
