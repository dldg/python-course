```python
# This is a class
class Dog():
    
    # Class Object Level Attribute (C++ eq. static)
    # Same for any instance of a class
    species = 'mammal'
    
    # This is a Constructor
    def __init__(self,breed,name,spots):
        # Attributes
        # We take in the argument
        # Assign it using self.attribute_name
        # Conventions is that both the argument and
        # the attribute have the same name
        self.breed = breed
        self.name = name
        # Expect boolean True/False
        self.spots = spots
        
    # Operations/Actions ----> Methods
    def bark(self,number):
        print("WOOF! My name is {} and the number is {}".format(self.name,number))
```

```python
my_dog = Dog(breed='Huskie',name='Sammy',spots=False)
```

```python
type(my_dog)
```

```python
my_dog.species
```

```python
my_dog.bark(2)
```

```python
class Circle():
    # Class Object Attributes
    pi = 3.14
    
    # Default value for arguments
    def __init__(self,radius=1):
        self.radius = radius
        # Class Object Attribute reference with the name of the class
        self.area = radius*radius*Circle.pi
    
    # Method
    def get_circumference(self):
        return self.radius * self.pi * 2
```

```python
my_circle = Circle(30)
```

```python
my_circle.pi
```

```python
my_circle.radius
```

```python
my_circle.get_circumference()
```

```python
my_circle.area
```
