```python
# could have been represented as list
# game_board = [' ']*10
game_board = [['7','8','9'], ['4','5','6'], ['1','2','3']]
```

```python
locations_dict = {'1': [2, 0],
                    '2': [2, 1],
                    '3': [2, 2],
                    '4': [1, 0],
                    '5': [1, 1],
                    '6': [1, 2],
                    '7': [0, 0],
                    '8': [0, 1],
                    '9': [0, 2]}
```

```python
def print_welcome():
    print("Welcome to Tic Tac Toe!!")
```

```python
from IPython.display import clear_output

def display_game():
    '''
    OUTPUT as follows
    |7|8|9|
    |4|5|6|
    |1|2|3|
    '''
    # will clear the board instead of
    # seeing the history of the changes!!
    clear_output()
    for row in game_board:
        for idx, cell in enumerate(row):
            if idx % 2 == 0:
                print("|" + cell, end = "|")
            else:
                print(cell, end = "")
        print()
```

```python
def clear_board():
    global game_board
    game_board = [['7','8','9'], ['4','5','6'], ['1','2','3']]
```

```python
def user_choice():
    '''
    OUTPUT - the choice as string
    '''
    # VARIABLES
    
    # Initial
    choice = 'WRONG'
    acceptable_range = range(1,10)
    within_range = False
    
    while choice.isdigit() == False or within_range == False:
        choice = input("Please choose location (1-9) ")
        
        # DIGIT CHECK
        if choice.isdigit()==False:
            print("Sorry that is not a digit!")
        # RANGE CHECK
        elif int(choice) not in acceptable_range:
            print("Sorry, location not in range!")
            within_range = False
        # AVAILABILITY CHECK
        elif game_board[locations_dict[choice][0]][locations_dict[choice][1]] in ['X','O']:
            print("Location already taken!")
        else:
            within_range = True

    return choice
```

```python
def handle_turn(letter):
    print(letter + "'s turn!")
    choice = user_choice()
    game_board[locations_dict[choice][0]][locations_dict[choice][1]] = letter
```

```python
def check_for_win(letter):
    '''
    OUTPUT - boolean whethere there was a win or not.
    '''
    win = False
    # check rows have same letters
    if game_board[0][0] == game_board[0][1] == game_board[0][2] == letter:
        win = True
    if game_board[1][0] == game_board[1][1] == game_board[1][2] == letter:
        win = True
    if game_board[2][0] == game_board[2][1] == game_board[2][2] == letter:
        win = True
    # check columns have same letters
    if game_board[0][0] == game_board[1][0] == game_board[2][0] == letter:
        win = True
    if game_board[0][1] == game_board[1][1] == game_board[2][1] == letter:
        win = True
    if game_board[0][2] == game_board[1][2] == game_board[2][2] == letter:
        win = True
    # check diagonals
    if game_board[0][0] == game_board[1][1] == game_board[2][2] == letter:
        win = True
    if game_board[2][0] == game_board[1][1] == game_board[0][2] == letter:
        win = True
        
    if win == True:
        print(letter + " is a winner!")

    return win
```

```python
def gameon_choice():
    '''
    OUTPUT - boolean whether to proceed with the game or not.
    '''
    choice = 'wrong'
    while choice not in ['Y','N']:
        choice = input("Keep Playing? (Y or N) ")
        if choice not in ['Y','N']:
            print("Sorry, I don't understand, please choose Y or N")
    if choice == "Y":
        return True
    else:
        return False
```

```python
import random

def choose_first():
    flip = random.randint(0,1)
    if flip == 0:
        return 'X'
    else:
        return 'O'
```

```python
game_on = True
turn = 'X'
number_of_turns = 0

clear_board()
# Randomally choose who starts.
turn = choose_first()
print_welcome()

while game_on:
    display_game()
    handle_turn(turn)
    number_of_turns = number_of_turns + 1
    win = check_for_win(turn)
    if win == True or number_of_turns == 9:
        clear_board()
        number_of_turns = 0
        game_on = gameon_choice()
        # A tie, set variables to default.
        if number_of_turns == 9:
            print("It's a tie!")
            turn = choose_first()
    else:
        turn = switch_turn(turn)
```
